```python
import pandas as pd
import json
```

```python
df = pd.read_json("extracted_conclusions/agg_df.json")
```

```python
parsed_dilemmas = json.load(open("ApproxDilemmasSegmentLocation.json", "r"))
```

```python
df
```

```python
df.iloc[0]["output"]
```

```python
df["len_acc"] = df["acceptable"].apply(lambda x: len(x))
```

```python
issues_df = df[df["len_acc"] == 0]
```

```python
issues_df
```

```python
for _, row in issues_df.iterrows():
    print(row["filename"])
    print(row["dilemma"])
    print(parsed_dilemmas[row["filename"]][row["dilemma"]]["selected"])
    print("-"*15)
```

```python
no_issue_df = df[df["len_acc"] > 0]
```

```python
no_issue_df.shape
```

```python
def viz_random_sample(df):
    sample = df.sample(1).iloc[0]
    for k in ["filename", "dilemma", "acceptable", "not_acceptable"]:
        v = sample[k]
        print(f"{k}: {v}")
        print("-"*15)

    print(parsed_dilemmas[sample["filename"]][sample["dilemma"]]["selected"])
    
```

```python
viz_random_sample(no_issue_df)
```

```python
no_issue_df.to_json("extracted_conclusions/agg_df_filtered.json")
```
